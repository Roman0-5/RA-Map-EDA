```python
import pandas as pd
from src.visualizing_pipeline      import visualize_raw
from src.log2_visualizing_pipeline import visualize_log

clinical  = pd.read_parquet("../../cleaned_datasets/clinical_merged.parquet")


# Expression: Patient_ID aus Index ableiten
expr_bl = pd.read_parquet("../../cleaned_datasets/exp_bl.parquet")
expr_bl['Patient_ID'] = expr_bl.index.str.replace('_BL', '', regex=False)
expr_bl = expr_bl.reset_index(drop=True)

expr_m6 = pd.read_parquet('../../cleaned_datasets/exp_m6.parquet')
expr_m6['Patient_ID'] = expr_m6.index.str.replace('_M6', '', regex=False)
expr_m6 = expr_m6.reset_index(drop=True)

pg_meta  = pd.read_parquet("../../cleaned_datasets/protogen_metadata.parquet")
protogen = pd.read_parquet("../../cleaned_datasets/protogen_features.parquet")

# BL
bl_idx     = pg_meta[pg_meta['Timepoint'] == 'BL'].index
pg_bl_meta = pg_meta.loc[bl_idx, ['Patient_ID']].reset_index(drop=True)
pg_bl_feat = protogen.loc[bl_idx].reset_index(drop=True)
pg_bl      = pd.concat([pg_bl_meta, pg_bl_feat], axis=1)
pg_bl      = pg_bl.rename(columns={c: f"{c}_BL" for c in pg_bl_feat.columns})

# M6
m6_idx     = pg_meta[pg_meta['Timepoint'] == 'M6'].index
pg_m6_meta = pg_meta.loc[m6_idx, ['Patient_ID']].reset_index(drop=True)
pg_m6_feat = protogen.loc[m6_idx].reset_index(drop=True)
pg_m6      = pd.concat([pg_m6_meta, pg_m6_feat], axis=1)
pg_m6      = pg_m6.rename(columns={c: f"{c}_M6" for c in pg_m6_feat.columns})

label_types = ['remission', 'eular', 'inflammation', 'therapy_x', 'therapy_y']
```

# Baseline Visualizations
Protgen + Proteomic

```python
for df, name, modality in [
    (expr_bl,  'expression_bl',  'expr'),
    (expr_m6,  'expression_m6',  'expr'),
    (pg_bl,    'protogen_bl',    'pg'),
    (pg_m6,    'protogen_m6',    'pg'),
]:
    visualize_raw(df, name=name, modality=modality,
                  clinical_df=clinical, label_type=label_types)
    visualize_log(df, name=name, modality=modality,
                  clinical_df=clinical, label_type=label_types)
```

```python

# --- Expression Baseline: Remission-Label ---
visualize_log(expr_bl, name="expression_bl",
              clinical_df=clinical, label_type='remission')

# --- Expression Baseline: EULAR-Label ---
visualize_log(expr_bl, name="expression_bl",
              clinical_df=clinical, label_type='eular')

# --- Protogen (merged hat Patient_ID + alle 163 Features) ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='eular')

visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='eular')

# --- Protogen: Therapy X ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_x')

# --- Protogen: Therapy X ---
visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_x')

# --- Protogen: Therapy Y ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_y')

# --- Protogen: Therapy Y ---
visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_y')

# --- Protogen: Inflammation ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='inflammation')

# --- Protogen: Inflammation ---
visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='inflammation')
```
