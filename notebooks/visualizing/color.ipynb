```python
import pandas as pd
from src.visualizing_pipeline      import visualize_raw
from src.log2_visualizing_pipeline import visualize_log

clinical  = pd.read_parquet("../../cleaned_datasets/clinical_merged.parquet")


# Expression: Patient_ID aus Index ableiten
expr_bl = pd.read_parquet("../../cleaned_datasets/exp_bl.parquet")
expr_bl['Patient_ID'] = expr_bl.index.str.replace('_BL', '', regex=False)
expr_bl = expr_bl.reset_index(drop=True)

# Protogen: features statt metadata laden
protogen = pd.read_parquet("../../cleaned_datasets/protogen_features.parquet")
# Patient_ID aus metadata dazuholen
pg_meta  = pd.read_parquet("../../cleaned_datasets/protogen_metadata.parquet")
pg_bl    = pg_meta[pg_meta['Timepoint'] == 'BL'][['Patient_ID']].reset_index(drop=True)
protogen = pd.concat([pg_bl, protogen], axis=1)

# --- Expression Baseline: Remission-Label ---
visualize_log(expr_bl, name="expression_bl remission",
              clinical_df=clinical, label_type='remission')

# --- Expression Baseline: EULAR-Label ---
visualize_log(expr_bl, name="expression_bl eular",
              clinical_df=clinical, label_type='eular')

# --- Protogen (merged hat Patient_ID + alle 163 Features) ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='eular')

visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='eular')

# --- Protogen: Therapy X ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_x')

# --- Protogen: Therapy X ---
visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_x')

# --- Protogen: Therapy Y ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_y')

# --- Protogen: Therapy Y ---
visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='therapy_y')

# --- Protogen: Inflammation ---
visualize_raw(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='inflammation')

# --- Protogen: Inflammation ---
visualize_log(protogen, name="protogen_bl",
              clinical_df=clinical, label_type='inflammation')
```
